```
Plot[{Sin[x], Cos[x], x/5}, {x, -2 Pi, 2 Pi},
 PlotLegends -> "Expressions", PlotStyle -> {Red, Blue, Green}]
```

```
ParametricPlot[{Cos[t], Sin[t]}, {t, 0, 2 Pi},
 PlotStyle -> Thick, AxesLabel -> {"x", "y"}]
```

```
data = Table[{x, Sin[x] + RandomReal[{-0.2, 0.2}]}, {x, 0, 10, 0.2}];
ListPlot[data, PlotStyle -> Blue, Filling -> Axis]
```
